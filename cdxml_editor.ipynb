# CDXML editor

Upload a ChemDraw CDXML file and choose a structure when the document contains several sketches. The PNG preview appears automatically. Toggling **Clean up 2D carbon geometry** immediately shows the cleaned sketch; unticking it restores the original drawing. **Create model** uses the geometry shown in that preview.

For crowded finite molecules, **Resolve steric contacts in 3D** rotates connected aromatic blocks and methyl hydrogens while preserving the chemical graph. **3D method** offers **Steric separation** (75% of summed van der Waals radii) or **Minimal 3D torsion** (editable C–C, C–H and H–H minimum distances, plus **Extra margin (%)**). The sketch preview stays 2D; **Create model** produces a 3D starting conformer. This is not an energy optimization; minimal mode can retain steric overlaps. Untick the option to keep the planar geometry.

Cleaned repeat brackets extend one C–C bond above and below the displayed atoms. Adjust **Left (Å)** and **Right (Å)** independently with the arrows or offset fields; the step starts at **0.1 Å**, and each bracket has its own reset. These drawing shifts leave the model and its periodic cell unchanged.

The preview preserves its aspect ratio with a maximum side of 300 pixels. Use **Enlarge preview** to inspect details.

Review and edit the result below. For square-bracketed repeat units, use **Infinite** for a periodic ribbon or enter a number of finite units. The **Download** tab includes **Extended xyz**, preserving the cell and periodicity.

Set a label and description, then click **Store to AiiDA database**. The confirmation below the editor shows the PK of the saved structure.

```python
import aiidalab_widgets_base as awb
import aiidalab_widgets_empa as awe
from aiida import load_profile
from aiidalab_widgets_base.viewers import StructureDataViewer
from IPython.display import display

load_profile()
```

```python
cdxml_importer = awe.CdxmlUploadWidget(title="CDXML")

editor = awb.StructureManagerWidget(
    importers=[cdxml_importer],
    editors=[
        awb.BasicStructureEditor(title="Edit structure"),
        awb.BasicCellEditor(title="Edit cell"),
    ],
    node_class="StructureData",
    storable=True,
    viewer=StructureDataViewer(
        configuration_tabs=["Cell", "Selection", "Appearance", "Download"],
    ),
)
editor.btn_store.description = "Store to AiiDA database"
editor.btn_store.layout.width = "auto"
display(editor)
```
